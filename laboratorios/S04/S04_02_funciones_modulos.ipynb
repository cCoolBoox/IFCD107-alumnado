## Contexto TurisData
En el cuaderno anterior calculamos importes reserva a reserva. Si TurisData tiene 6.000 reservas no vamos a
copiar y pegar el código 6.000 veces: lo convertimos en **funciones reutilizables** que luego usaremos en el reto.

```python
# Preparación (ejecuta esta celda siempre la primera)
import os, urllib.request

BASE_URL = "https://raw.githubusercontent.com/cCoolBoox/IFCD107-alumnado/main/datos/"

def dato(nombre):
    """Devuelve la ruta local de un archivo de datos del curso (lo descarga si no está)."""
    for carpeta in ("datos", "../datos", "../../datos"):
        ruta = os.path.join(carpeta, nombre)
        if os.path.exists(ruta):
            return ruta
    os.makedirs("datos", exist_ok=True)
    ruta = os.path.join("datos", nombre)
    urllib.request.urlretrieve(BASE_URL + nombre, ruta)
    return ruta
```

## 1 · Definir una función
Una función recibe **parámetros** (entradas), hace algo y devuelve un resultado con `return`. Sin `return`
devuelve `None`. La primera cadena entre triples comillas es el **docstring** (documentación).

```python
def ingreso_reserva(noches, precio_noche):
    """Devuelve el ingreso bruto de una reserva (noches por precio de la noche)."""
    return noches * precio_noche

print(ingreso_reserva(3, 89.24))
help(ingreso_reserva)     # muestra el docstring
```

**Parámetros por defecto y con nombre.** Un parámetro con valor por defecto es opcional; puedes pasarlos por
nombre para que la llamada se lea mejor.

```python
def ingreso_con_tasa(noches, precio_noche, tasa_por_noche=1.5):
    """Ingreso de la reserva incluyendo la tasa turística por noche (1,5 EUR por defecto)."""
    return noches * (precio_noche + tasa_por_noche)

print(ingreso_con_tasa(4, 80))                       # usa la tasa por defecto
print(ingreso_con_tasa(4, 80, tasa_por_noche=0))     # sin tasa
```

## 2 · `*args`, `**kwargs` y `lambda`
- `*args` recoge **cualquier número de valores** en una tupla.
- `**kwargs` recoge **parámetros con nombre** en un diccionario.
- `lambda` crea una función corta sin nombre; muy útil como criterio de ordenación.

```python
def media(*valores):
    """Media de todos los valores recibidos."""
    return sum(valores) / len(valores)

def ficha(**datos):
    """Devuelve una línea con los datos recibidos como pares clave=valor."""
    return ", ".join(f"{k}={v}" for k, v in datos.items())

print(media(60, 72.5, 89.24))
print(ficha(canal="directo", noches=3))

reservas = [("A", 3, 80.0), ("B", 5, 60.0), ("C", 2, 150.0)]
print(sorted(reservas, key=lambda r: r[1] * r[2], reverse=True))    # ordenar por ingreso
```

## 3 · Módulos: importar código ya escrito
Un **módulo** es un archivo `.py`. Python trae muchos ("la biblioteca estándar"): `math`, `statistics`,
`random`, `datetime`, `csv`, `json`... Se importan con `import`.

```python
import math
import statistics
from datetime import date

print(math.sqrt(16), math.ceil(4.2))
print(statistics.mean([60, 72.5, 89.24]), statistics.median([60, 72.5, 89.24]))

llegada = date(2025, 12, 24)
salida = date(2025, 12, 28)
print("Noches:", (salida - llegada).days)
```

## 4 · Crear tu propio módulo
En un proyecto real las funciones útiles viven en un archivo (`kpis.py`, `utilidades.py`...) y se importan.
Aquí lo simulamos: escribimos un módulo en una carpeta temporal y lo importamos.

```python
import importlib
import os
import sys
import tempfile

carpeta = tempfile.mkdtemp()
with open(os.path.join(carpeta, "utilidades_turisdata.py"), "w", encoding="utf8") as f:
    f.write('''"""Utilidades de TurisData."""

def ingreso(noches, precio):
    """Ingreso bruto de una reserva."""
    return noches * precio

def pct(parte, total):
    """Porcentaje con 1 decimal; 0.0 si el total es 0."""
    return round(100 * parte / total, 1) if total else 0.0
''')
sys.path.insert(0, carpeta)                 # para que Python encuentre el módulo
utilidades_turisdata = importlib.import_module("utilidades_turisdata")
print(utilidades_turisdata.ingreso(3, 80), utilidades_turisdata.pct(2754, 6000))
```

Reglas de estilo (PEP 8) que seguiremos: nombres en `minusculas_con_guion_bajo`, 4 espacios de sangría,
docstring en cada función, líneas de menos de ~100 caracteres.
