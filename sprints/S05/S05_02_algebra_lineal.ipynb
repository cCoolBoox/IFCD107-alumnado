## 1 · Vectores
Un **vector** es una lista de números con un significado: por ejemplo, las noches vendidas de cada tipo de
habitación. Se pueden sumar, multiplicar por un número y medir.
El **producto escalar** (`np.dot` o `@`) multiplica elemento a elemento y suma: es exactamente
"noches por precio, sumado".

```python
noches = np.array([10, 4, 2])                 # estándar, superior, suite
precios = np.array([70.0, 95.0, 160.0])       # EUR por noche de cada tipo
print("ingreso:", noches @ precios)           # 10x70 + 4x95 + 2x160
print("suma de vectores:", noches + np.array([1, 1, 1]), "· por un escalar:", noches * 2)
print("norma (longitud):", np.linalg.norm(noches).round(3))
```

## 3 · Sistemas de ecuaciones
Si desconocemos los precios pero conocemos varias facturas, obtenemos un **sistema**: `A x = b`, donde `A` guarda
las noches vendidas, `x` son los precios desconocidos y `b` los importes facturados. `np.linalg.solve(A, b)` lo resuelve.

```python
# Dos días de facturación con solo dos tipos de habitación:
A = np.array([[10, 4],
              [ 8, 6]], dtype=float)
b = np.array([1180.0, 1240.0])
x = np.linalg.solve(A, b)
print("precios:", x, "· comprobación:", A @ x)
```

## 4 · Inversa e identidad
La **matriz inversa** `A⁻¹` deshace a `A`: `A @ A⁻¹` es la **identidad** (unos en la diagonal, ceros fuera).
No todas las matrices tienen inversa (si el determinante es 0, no). En la práctica usamos `solve` en lugar de calcular la inversa.

```python
inversa = np.linalg.inv(A)
print(np.round(A @ inversa, 10))
print("determinante:", np.linalg.det(A).round(2))
```

## 5 · Álgebra lineal y regresión: la ecuación normal
Una **regresión lineal** predice `y ≈ X @ w`. `X` guarda las variables (con una columna de unos para el término
independiente) y `w` los coeficientes. Los coeficientes que minimizan el error cuadrático cumplen la **ecuación normal**:

`(Xᵀ X) w = Xᵀ y`      →      `w = np.linalg.solve(X.T @ X, X.T @ y)`

Es un sistema de ecuaciones como el de antes. Lo probamos con los datos de TurisData: ocupación según temperatura.
