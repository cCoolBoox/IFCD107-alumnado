## 5 · Repetir: `for` y `while`
`for` recorre una secuencia; `range(n)` produce 0, 1, ..., n-1. `while` repite mientras se cumpla una condición
(cuidado con los bucles infinitos: siempre debe cambiar algo dentro).

```python
for noche in range(1, 4):
    print("Noche", noche)

ocupadas = 0
while ocupadas < 3:
    ocupadas += 1          # forma corta de ocupadas = ocupadas + 1
print("Habitaciones ocupadas:", ocupadas)
```

```python
#@title 🆘 Solución de referencia (si este ejercicio se queda para casa, ejecuta esta celda para seguir; ábrela solo después de intentarlo)
presupuesto = 500
precio = 89
noches_pagables = 0
restante = presupuesto
while restante >= precio:
    restante -= precio
    noches_pagables += 1
assert noches_pagables == 5, f"500 // 89 = 5 noches; tienes {noches_pagables}"
print("Correcto:", noches_pagables)
```
