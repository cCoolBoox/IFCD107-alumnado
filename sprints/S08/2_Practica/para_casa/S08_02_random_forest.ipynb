## 1. La idea: bagging + variables al azar

#### A2 · Ejemplo — Qué reservas se quedan fuera de una muestra bootstrap

Un bosque construye muchos árboles y **promedia** sus predicciones. Para que los árboles sean *distintos entre sí* (si fueran iguales, promediar no serviría), se introduce aleatoriedad de dos formas:
1. **Bagging** (*bootstrap aggregating*): cada árbol se entrena con una **muestra con reemplazo** de las reservas (unas aparecen repetidas y otras, ≈ 37 %, no aparecen).
2. En cada pregunta, cada árbol solo puede elegir entre un **subconjunto aleatorio de variables** (`max_features`).

Cada árbol individual es peor que uno bien podado, pero **sus errores no coinciden**; al votar, el ruido se cancela y baja la varianza. Las reservas que un árbol no vio (**out-of-bag**, OOB) sirven para evaluar sin necesidad de validación cruzada.

- **Qué hace:**
    - `rng.integers(0, n, n)` hace una muestra **con reemplazo** de 5000 posiciones sobre 5000 reservas (algunas salen varias veces).
    - `set(range(n)) - set(...)` son las reservas que **no** salieron; dividido entre `n` da la proporción.
- **Qué debe salir:**
    - `Reservas que NO entran en una muestra bootstrap: 37.1%`.
- **Cómo leerlo:**
    - Es el 37 % (≈ 1/e = 36,8 %) que cada árbol no ve: esas reservas «fuera de la bolsa» (*out-of-bag*) sirven de validación gratuita.
- **Si falla:**
    - `NameError: name 'np' is not defined` → falta ejecutar la celda de preparación.

2. *¿Qué quiere decir muestreo «con reemplazo»?* → Que tras sacar una reserva se **devuelve** al total: puede salir varias veces, y por eso algunas no salen ninguna.

## 2. Cuántos árboles y de qué tamaño

- `n_estimators`: **más árboles casi nunca perjudica** (solo cuesta tiempo); se estabiliza.

## 3. Importancia de variables: dos formas de medirla

#### A6 · Ejemplo — Importancia por permutación en el pipeline completo

- **Importancia de impureza** (`feature_importances_`): rápida, pero sesgada hacia variables numéricas con muchos valores distintos (precio, antelación) y calculada **sobre train**.
- **Importancia por permutación**: se **baraja al azar** una columna del **test** y se mide cuánto empeora el AUC. Si empeora mucho, el modelo dependía de esa variable. Más fiable, pero más lenta.

Para poder leerla en columnas *originales* (no en cada one-hot), aplicamos la permutación a un `Pipeline` completo.

- **Qué hace:**
    - `modelo_bosque` es un `Pipeline` (preprocesado + bosque de 300 árboles) entrenado con `X_train` (las columnas originales).
    - `permutation_importance(..., scoring="roc_auc", n_repeats=5)` baraja cada columna de `X_test` 5 veces y mide cuánto cae el AUC; `imp_perm` es la caída media, ordenada de mayor a menor.
    - Dibuja el resultado en barras horizontales con su desviación (`xerr`).
- **Qué debe salir:**
    - Un gráfico con las 15 variables. Arriba: `antelacion_dias` (caída ≈ 0,110), `deposito` (≈ 0,047), `canal` (≈ 0,043), `tarifa` (≈ 0,031). Abajo, varias con valor ≈ 0 (algunas ligeramente negativas).
- **Cómo leerlo:**
    - Un valor **negativo** (p. ej. `pais_origen`, −0,0015) significa que barajar esa variable no empeora nada: es ruido, **no aporta**.
- **Si falla:**
    - `NameError: name 'prep' is not defined` → falta ejecutar la celda de preparación.

## 6. Para reflexionar

1. ¿Por qué promediar árboles **distintos** reduce el error, y por qué no serviría promediar árboles idénticos?
2. ¿Qué diferencia hay entre `feature_importances_` y la importancia por permutación? ¿Cuál usarías para explicar al cliente?
3. ¿Cuándo usarías la puntuación OOB en lugar de validación cruzada?
4. Si el bosque solo mejora 0,005 de AUC a la logística, pero es 50 veces más lento, ¿qué decides?

<details><summary>Ver respuestas (después de pensarlo)</summary>

1. Porque los errores de árboles distintos **no coinciden** y al votar se cancelan (baja la varianza). Con árboles idénticos todos darían la misma respuesta y el promedio sería igual que uno solo.
2. `feature_importances_` se calcula con **train** y favorece a las numéricas con muchos valores distintos. La permutación se calcula en **test**, barajando cada variable y midiendo cuánto cae el AUC: es más fiable. Para explicar al cliente usaría la de **permutación** («si no tuviéramos esta variable, el modelo empeoraría X»).
3. Cuando entrenar es caro o se quiere ver la evolución con el nº de árboles: el OOB sale **gratis** del propio entrenamiento. Con pocos datos o para una comparación más rigurosa entre modelos, la validación cruzada es más completa.
4. Elegiría la **logística** (o el modelo simple): el rendimiento es casi el mismo, 50 veces más rápido, más barato de mantener y más explicable. En nuestros datos, además, la logística (0,758) ganó al bosque (0,741).

</details>

## 7. Resumen: qué has aprendido
- Bosque = muchos árboles con **bagging** y **variables al azar**, promediados.
- Más árboles no sobreajusta; se ajustan `min_samples_leaf`, `max_depth`, `max_features`.
- OOB da una validación «gratis»; la permutación da una importancia más fiable.
- Rendimiento, explicabilidad y coste van en tensión: hay que compararlos.
